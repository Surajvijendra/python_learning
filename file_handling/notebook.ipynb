### Def function 

```python
def add_numbers(a,b):
    c = a+b
    print(c)
    return c
```

```python
add_numbers(2,3)
```

```python
# positional argument 
# will be used to add multipule values want to aded 
def team(*args):
    team1 = []
    team2 = []
    for name in args:
        if name != 'Brett Lee':
            team1.append(name)
        elif name == 'Brett Lee':
            team2.append(name)
    return print(f'team 1 will be indians{team1}, team 2 will be non indians{team2}')
```

```python
team('Bhuvi','Dhoni','Rahane','Yuvraj','Brett Lee')
```

```python
def greeting(name,jercy):
    return f'hello {name} welcome to {jercy}'
```

```python
def palan(data):
    data = data.lower().replace(" ",'')
    if data == data[::-1]:
        return print(f'{data} is palandrom')
    else:
        return print(f'{data} not a palandrom')
```

```python
palan('aba')
```
