### Functions
* A function is a block of code that only runs when it is called. You can pass data (called parameters) into it, and it can return data using return.

```python
def greeting(name):
    return f'hello {name}'
```

```python
greeting('suraj')
```

```python
def even_or_odd(n):
    if n % 2==0:
        print(f'{n} --> is an even number')
    else:
        print(f'{n} --> is a odd number')
```

```python
even_or_odd(8)
```

```python
even_or_odd(7)
```

```python
def factl(n):
    values = 1
    for i in range(1,(n+1)):
        values *= i
    return (values)
```

```python
factl(6)
```

```python
factl(4)
```

```python
def factorial_number(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial_number(n - 1)

print(factorial_number(5))  # Output: 120
```

### Return Multiple values

```python
def mult_math(a,b):
    return a+b,a-b
```

```python
addition_,subtraction_ = mult_math(2,5)
```

```python
print(addition_)
print(subtraction_)
```

### *args: Variable Number of Positional Arguments
* Use *args when you don’t know how many positional arguments will be passed.

```python
def multi_values(*args):
    return (args)
```

```python
multi_values(5,3)
```

```python
def multi_values(*args):
    return (sum(args))
```

```python
multi_values(1,2,3,4)
```

### ** args Variable Number of Keyword Arguments

```python
def key_value_pair(**args):
    return args
```

```python
key_value_pair(name='Bhuvi',profile='cricketer')
```

```python
def key_value_pair(**args):
    for key,values in args.items():
        print(f'{key} --> {values}')
```

* if i mention return it will provide the last saved value
* if i use print then i will get all the values

```python
key_value_pair(name='Bhuvi',profile='cricketer')
```
