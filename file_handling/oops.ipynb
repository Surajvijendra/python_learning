```python
class gamer:
    def __init__(self,game_name,year_exp):
        self.game_name = game_name
        self.year_exp = year_exp
        
    def bgmi(self):
        return f'your are playing this game {self.game_name} from last {self.year_exp}'
```

```python
gamer('BGMI',5).bgmi()
```

```python
class Gamer:
    def __init__(self):
        self.games = []  # empty list to store (game_name, year_exp)

    def add_game(self, game_name, year_exp):
        self.games.append((game_name, year_exp))  # add as tuple

    def show_games(self):
        for game, exp in self.games:
            print(f"You are playing {game} for the last {exp} years.")
```

```python
g = Gamer()
g.add_game("BGMI", 6)
g.add_game("Clash Royal", 8)
g.add_game("Call of Duty", 4)

g.show_games()
```
